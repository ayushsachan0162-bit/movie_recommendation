```python
import numpy as np
import pandas as pd
```

```python
movies= pd.read_csv("tmdb_5000_movies.csv")
credits=pd.read_csv('tmdb_5000_credits.csv')
movies.shape
```

```python
## merging both data set
movies=movies.merge(credits, on='title')
```

```python
# include all the columns which is needed
movies=movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
```

```python
#checking the null values
movies.isnull().sum()
```

```python
## delete the rows of overview which is null
movies.dropna(inplace=True)
```

```python
## checking duplicate data
movies.duplicated().sum()
```

```python
movies.iloc[0].genres
```

```python
movies.head()
```

```python
movies.head(1)
```

```python
import ast
def convert1(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter!=3:
            L.append(i['name'])
            counter=counter+1
        else:
            break
            
       
    return L
```

```python
movies['cast']=movies['cast'].apply(convert1)
```

```python
movies.head(1)
```

```python
import ast
def convert2(obj):
    L=[]
 
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L
```

```python
movies['crew']=movies['crew'].apply(convert2)
```

```python
movies.head(1)
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.head(1)
```

```python
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ", "") for i in x])
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ", "") for i in x])
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ", "") for i in x])
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ", "") for i in x])
```

```python
movies.head(1)
```

```python
movies['tags']=movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
movies.head()
```

```python
movies['tags'][0]
```

```python
new_df=movies[['movie_id', 'title', 'tags']]
new_df
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df.head()
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:x.lower())
```

```python
new_df
```

```python
## PRE PROCESS IS DONE 
```

```python
## NOW WE USE bAGS OF WORDS METHOD FOR TEXT VECTORIZATION
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000, stop_words="english")
```

```python
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors
```

```python
for  word in cv.get_feature_names_out():
  print(word)
```

```python
import nltk ## it is a natural language toolkit
from nltk.stem.porter import PorterStemmer
ps= PorterStemmer()
```

```python
ps.stem('loved')
```

```python
def stem(text):## doubt 4
    y=[]

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

```python
new_df
```

```python
## distance is inversely proportional to similarity
## we just calculate angles between two vectors(film)
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_similarity(vectors)
```

```python
similarity=cosine_similarity(vectors)
```

```python
similarity[1]
```

```python
sorted(similarity[0])[-10:-1]
```

```python
list(enumerate(similarity[0]))
```

```python
#index=0 ke pass me jo movie hai , uska index show below
sorted(list(enumerate(similarity[0])), reverse=True, key=lambda x:x[1])[1:6]
```

```python
def recommend(movie): ## doubt5
    movie_index=new_df[new_df['title']==movie].index[0]
    distances=similarity[movie_index]
    movies_list=sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Avatar')
```
